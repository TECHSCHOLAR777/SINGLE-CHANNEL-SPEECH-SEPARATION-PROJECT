```python
cmd = [
    sys.executable, 'train/stage1_single.py',
    '--adapter',           'noise',
    '--data-root',         'data/calmsep-8k',
    '--checkpoint-dir',    'checkpoints/stage1_noise',
    '--device',            'cuda',
    '--batch-size',        '8',
    '--epochs',            '40',
    '--samples-per-epoch', '500',
    '--lr',                '1e-4',
    '--max-clip-samples',  '16000',
    '--num-workers',       '2',
]
proc = subprocess.Popen(cmd, env=os.environ.copy(),
                        stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
                        text=True, bufsize=1)
for line in proc.stdout:
    print(line, end='')
proc.wait()
print('exit:', proc.returncode)
```

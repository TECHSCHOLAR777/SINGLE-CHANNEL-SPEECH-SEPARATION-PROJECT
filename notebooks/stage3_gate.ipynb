# CoRAL-Sep Stage 3: Gate + Level-2 Analyzer Training

Train GateNetwork and Level2Analyzer jointly.

Prerequisites:
- Stage 1 adapters (reverb, noise, codec) in `STAGE1_DIR`.
- Data manifests at `DATA_ROOT`.

**Held-out combos** `reverb+codec` and `noise+codec` are excluded from
gate training (BLUEPRINT §7.5) — confirmed via `assert_not_held_out()`.

Expected time: ~4 hours on T4.

```python
DATA_ROOT      = "/kaggle/input/calmsep-8k"
CHECKPOINT_DIR = "/kaggle/working/gate"
STAGE1_DIR     = "/kaggle/working/adapters"
EPOCHS         = 30
BATCH_SIZE     = 8
LR             = 5e-5
DEVICE         = "cuda"
REPO_PATH      = "/kaggle/input/calmsep-code"
NOISE_DIR      = "/kaggle/input/calmsep-8k/noise"
RIR_BANK       = "/kaggle/input/calmsep-8k/rirs/bank.json"
BUT_DIR        = "/kaggle/input/calmsep-8k/but-reverbdb"
```

```python
import sys, subprocess
if REPO_PATH not in sys.path:
    sys.path.insert(0, REPO_PATH)

cmd = [
    sys.executable, "-m", "train.stage3_gate",
    "--data-root", DATA_ROOT,
    "--checkpoint-dir", CHECKPOINT_DIR,
    "--stage1-dir", STAGE1_DIR,
    "--epochs", str(EPOCHS),
    "--batch-size", str(BATCH_SIZE),
    "--lr", str(LR),
    "--device", DEVICE,
    "--noise-dir", NOISE_DIR,
    "--rir-bank",  RIR_BANK,
    "--bf16",
]
print("Running:", " ".join(cmd))
subprocess.run(cmd, cwd=REPO_PATH)
```

```python
import os
ckpts = [
    "gate_net.pt",
    "level2_analyzer.pt",
]
for ckpt in ckpts:
    path = os.path.join(CHECKPOINT_DIR, ckpt)
    print(f"{ckpt}: {'OK' if os.path.exists(path) else 'MISSING'}")
```
